Create a Constructor for a Person Write a JavaScript function constructor named Person that takes two parameters, name, and age, and Add a method to the prototype to display name and age. Then, create two instances of Person and display their names and ages.

**Problem Statement 2:**

Implement a Bank Account Create a constructor function named BankAccount that initializes a bank account with an initial balance. Include methods for depositing and withdrawing money from the account. Write code to demonstrate these operations on a bank account object.


```
function BankAccount(initialBalance) {
  this.balance = initialBalance;
  this.deposit = function(amount) {
    this.balance += amount;
    console.log(`Deposited: ${amount}, New Balance: ${this.balance}`);
  }
  this.withdraw = function(amount) {
    if(amount > this.balance){
      console.log("Insufficient funds");
    } else {
      this.balance -= amount;
      console.log(`Withdraw: ${amount}, New Balance: ${this.balance}`);
    }
  }
}
```

**TASK 5**

 **Problem Statement 5:**

Implement a Calculator Class Create a class called Calculator that initializes two values value1 and value2 to store numbers. Add methods for add, subtract, multiply, and divide. Perform and show operations.


```
class Calculator {
  constructor(num1,num2) {
    this.value1 = num1;
    this.value2 = num2;
  }
  add() {
    return this.value1 + this.value2;
  }
  substract() {
    return this.value1 - this.value2;

  }
  multiply() {
    return this.value1 * this.value2;

  }
  divide() {
    return this.value1 / this.value2;
  }
}
const calc = new Calculator(10, 5);
console.log("Addition: " + calc.add());
console.log("Substraction: " + calc.substract());
console.log("Multiplication: " + calc.multiply());
console.log("Division: " + calc.divide());
```

**Problem Statement 7:**

Implement a Library System Create a class called Library that can manage a collection of books. Each book should be represented as an instance of a Book class (with properties like title, author, and availability). Implement methods in the Library class to add books, check out books, and return books. Keep track of the books available.


```
class books {
  constructor(title, author) {
    this.title = title;
    this.author = author;
    this.available = true;
  }
}


class Library {
  constructor() {
    this.books = [];
  }

  addBook(book) {
    this.books.push(book);
  }

  checkOutBook(title) {
    const book = this.books.find(b => b.title === title);
    if (book && book.available) { // ky ye book class s a rh ? - yes - ok so why we can not use this keyword here - because this is library class method
      book.available = false; // book availability false kr diya gya kyuki checked out ho gya
      console.log(`Checked out: ${book.title}`);
    } else {
      console.log(`Book not available: ${title}`);
    }
  }

  returnBook(title) {
    const book = this.books.find(b => b.title === title); //return krega ki book mil gya ya nhi agar mil gya to uski availability true krdega and find method return matched object
    if (book) {
      book.available = true;
      console.log(`Returned: ${book.title}`);
    }
  }

  listAvailableBooks() {
    const availableBooks = this.books.filter(b => b.available); // its return array ofall available books
    console.log("Available books:");
    availableBooks.forEach(b => console.log(`- ${b.title} by ${b.author}`)); // for each available book print title and author

  }


}

const library = new Library(); // 1st we are creating library object because its create empty array ? -

const book_1= new books("atomic habits", "james clear");
const book_2 = new books("the power of habit", "charles duhigg");
library.addBook(book_1);
library.addBook(book_2);

library.listAvailableBooks();
library.checkOutBook("atomic habits");
library.listAvailableBooks();
library.returnBook("atomic habits");
library.listAvailableBooks();
```

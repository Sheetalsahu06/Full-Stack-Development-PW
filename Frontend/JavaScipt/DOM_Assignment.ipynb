**Q1. Explain the DOM and its role in Web development.**

What is DOM (Document Object Model)?

 * The DOM is a programming interface for HTML and XML documents.

 * It represents a web page as a tree of objects (nodes).

 * Each HTML element becomes an object that JavaScript can access and modify.

How DOM is Created

 * Browser loads HTML

 * Parses it

 * Converts it into a DOM tree

HTML:
```html
<h1 id="title">Hello</h1>
<p>Welcome</p>

```
JS:
```javascript
document.getElementById("title").textContent = "Hi User";

```
Role of DOM in Web Development

 * Access HTML elements

 * Modify content dynamically

 * Change styles

 * Add or remove elements

 * Handle user events

**Q2. Explain the concept of event delegation and provide a scenario where it is beneficial.**

What is Event Delegation?

 * Event delegation is a technique where a single event listener is attached to a parent element instead of multiple child elements.

 * It works because of event bubbling.

 Why Event Delegation is Used

 * Improves performance

 * Reduces memory usage

 * Handles dynamically added elements
```html
<ul id="list">
  <li>Item 1</li>
  <li>Item 2</li>
</ul>

```
```javascript
document.getElementById("list").addEventListener("click", function (e) {
  if (e.target.tagName === "LI") {
    console.log(e.target.textContent);
  }
});

```
Scenario Where It Is Beneficial

 * A to-do list app where new list items are added dynamically.

 * Instead of adding a click listener to every `<li>`, one listener on `<ul>` handles all.

**Q3. Explain the concept of Event Bubbling in the DOM.**

What is Event Bubbling?

 * Event bubbling is a process where an event starts from the target element and moves upward to its parent elements.

Why Event Bubbling is Important

 * Makes event delegation possible

 * Helps manage events efficiently

```html
<div id="parent">
  <button id="child">Click</button>
</div>
```
```javascript

document.getElementById("child").addEventListener("click", () => {
  console.log("Child clicked");
});

document.getElementById("parent").addEventListener("click", () => {
  console.log("Parent clicked");
});
```

**Output:**
```text
Child clicked
Parent clicked
```

**7. Build a dynamic dropdown menu using HTML and JavaScript. Create an HTML structure for a navigation menu with dropdowns. Use JavaScript to toggle the visibility of dropdowns when the user hovers over menu items.**

```html
<!DOCTYPE html>
<html lang="en">

<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
    <style>
        .dropdown-menu {
            display: none;
        }
      

    </style>
</head>

<body>
    <div id="menu">
        <ul>
            <li>
                <a href="#">Home</a>
            </li>
            <li class="dropdown">
                <a href="#">Services</a>
                <ul class="dropdown-menu">
                    <li><a href="#">Submenu4</a></li>
                    <li><a href="#">Submenu5</a></li>
                    <li><a href="#">Submenu6</a></li>
                </ul>
            </li>
            <li><a href="#">Contact</a></li>
        </ul>
    </div>
    <script>
        const dropdown = document.querySelector('.dropdown');
        const dropdownMenu = document.querySelector('.dropdown-menu');

        dropdown.addEventListener('mouseover', () => {
            dropdownMenu.style.display = 'block';
        });

        dropdown.addEventListener('mouseout', () => {
            dropdownMenu.style.display = 'none';
        });
    </script>
</body>

</html>
```

**8. Create a simple dynamic shopping list with the following features**
* The item should appear in the list
* Each item should be given a button that can be pressed to delete that item off the list
* The input should be emptied and focused ready for you to enter another item

```html
<!DOCTYPE html>
<html lang="en">

<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
    <style>
        #inputContainer {
            margin: 10px;
        }
        input,
        button {
            padding: 5px;

        }
        .list-item {
            display: flex;
            gap: 5px;
            margin: 5px;
        }
        li,
        button {
            padding: 5px;
        }
    </style>
</head>

<body>
    <div id="inputContainer">
        <input type="text" id="input" placeholder="Enter item">
        <button id="addItem">Add Item</button>
    </div>
    <div id="list-container">
        <ul>
            <div class="list-item">
                <li>Egg</li>
                <button class="delete-btn">Delete</button>
            </div>
            <div class="list-item">
                <li>Milk</li>
                <button class="delete-btn">Delete</button>
            </div>
            <div class="list-item">
                <li>Bread</li>
                <button class="delete-btn">Delete</button>
            </div>
        </ul>
    </div>
</body>
<script>
    const input = document.querySelector("#input");

    const addItem = document.querySelector("#addItem");

    addItem.addEventListener("click", (evt) => {
        const inputValue = input.value;
        console.log(inputValue);
        if (inputValue === "") {
            alert("please! Enter item name")
            return;
        }
        createListItem(inputValue);
        input.value = "";
    })

    function createListItem(inputValue) {
        const c = document.createElement("div");
        c.classList.add("list-item");

        const list = document.createElement("li");
        list.innerHTML = inputValue;

        const btn = document.createElement("button");
        btn.innerText = "Delete";
        btn.classList.add("delete-btn")


        c.append(list, btn);

        const ul = document.querySelector("ul");
        ul.appendChild(c)

    };

    document.querySelector("ul").addEventListener("click", (event) => {
        if (event.target.classList.contains("delete-btn")) {
            event.target.parentElement.remove();
        }
    });
</script>

</html>
```

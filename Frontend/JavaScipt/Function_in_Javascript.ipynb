**Q4. Write a JavaScript function called calculateTax that takes an income as an argument and returns the amount of tax to be paid. Use a closure to handle different tax rates based on income ranges. Test the function with various incomes.**

```
// Function that creates a tax calculator (closure)
function calculateTax() {
    // tax logic inside closure
    return function(income) {
        let tax;
        if (income <= 250000) {
            tax = 0;
        } else if (income <= 500000) {
            tax = (income - 250000) * 0.05;
        } else if (income <= 1000000) {
            tax = (250000 * 0.05) + (income - 500000) * 0.2;
        } else {
            tax = (250000 * 0.05) + (500000 * 0.2) + (income - 1000000) * 0.3;
        }
        return tax;
    }
}

// Create tax calculator
const taxCalculator = calculateTax();

// Testing with incomes
console.log("Income: 200000 => Tax: " + taxCalculator(200000));
console.log("Income: 400000 => Tax: " + taxCalculator(400000));
console.log("Income: 750000 => Tax: " + taxCalculator(750000));
console.log("Income: 1200000 => Tax: " + taxCalculator(1200000));


```

**5. Write a JavaScript function called factorial that calculates the factorial of a non-negative integer using recursion. Test the function with different inputs.**

```
// Recursive factorial function
function factorial(n) {
    // base case
    if (n === 0 || n === 1) {
        return 1;
    }
    // recursive case
    return n * factorial(n - 1);
}

// Testing the function
console.log(factorial(0));  // 1
console.log(factorial(1));  // 1
console.log(factorial(5));  // 120
```
